````python
# Check what's actually in the dataset folder
import os

dataset_path = r"D:\College\Major Project\datasets\ShanghaiTech"

print("📂 Exploring dataset contents...\n")

def list_directory_tree(path, prefix="", max_depth=3, current_depth=0):
    if current_depth >= max_depth:
        return
    
    try:
        items = sorted(os.listdir(path))
        for i, item in enumerate(items):
            item_path = os.path.join(path, item)
            is_last = i == len(items) - 1
            current_prefix = "└── " if is_last else "├── "
            print(f"{prefix}{current_prefix}{item}")
            
            if os.path.isdir(item_path) and current_depth < max_depth - 1:
                next_prefix = prefix + ("    " if is_last else "│   ")
                list_directory_tree(item_path, next_prefix, max_depth, current_depth + 1)
    except PermissionError:
        pass

list_directory_tree(dataset_path)
````
