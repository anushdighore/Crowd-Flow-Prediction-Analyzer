```python
# Press Shift+Enter to run this cell.
import cv2
import numpy as np
import requests
import urllib3

IP_CAM_URL = "https://192.168.1.6:8080/shot.jpg"

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

def fetch_frame(url: str, timeout: float = 5.0):
    try:
        response = requests.get(url, timeout=timeout, verify=False)
        response.raise_for_status()
        frame_array = np.frombuffer(response.content, dtype=np.uint8)
        frame = cv2.imdecode(frame_array, cv2.IMREAD_COLOR)
        return frame
    except requests.RequestException:
        return None

window_name = "Mobile IP Camera Stream"

print("Streaming from:", IP_CAM_URL)
print("Press 'q' in the window to exit.")

while True:
    frame = fetch_frame(IP_CAM_URL)
    if frame is None:
        print("⚠️  Failed to fetch frame, retrying...")
        if cv2.waitKey(500) & 0xFF == ord("q"):
            break
        continue

    cv2.imshow(window_name, frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cv2.destroyAllWindows()
```
