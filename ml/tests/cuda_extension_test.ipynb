```python
# Test 3: Functional Test with Real Tensors
import torch

print('='*70)
print('Functional Test: Selective Scan on GPU')
print('='*70)

if torch.cuda.is_available():
    try:
        from mamba_ssm.ops.selective_scan_interface import selective_scan_fn
        
        # Create test tensors
        batch = 2
        seq_len = 128
        d_inner = 64
        d_state = 16
        
        u = torch.randn(batch, d_inner, seq_len).cuda()
        delta = torch.randn(batch, d_inner, seq_len).cuda()
        A = torch.randn(d_inner, d_state).cuda()
        B = torch.randn(batch, d_state, seq_len).cuda()
        C = torch.randn(batch, d_state, seq_len).cuda()
        D = torch.randn(d_inner).cuda()
        
        # Run selective scan
        out = selective_scan_fn(
            u, delta, A, B, C, D,
            return_last_state=False
        )
        
        print(f'✅ Selective scan executed successfully')
        print(f'   Input shape: {u.shape}')
        print(f'   Output shape: {out.shape}')
        print(f'   Device: {out.device}')
        print()
        print('🚀 CUDA extensions are fully functional!')
        
    except Exception as e:
        print(f'❌ Functional test failed: {e}')
        print('\n   This means CUDA extensions are not working properly')
else:
    print('⚠️ CUDA not available - cannot test GPU operations')

print('='*70)
print()
```
