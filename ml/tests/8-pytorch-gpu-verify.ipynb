## 🎮 Step 2: Check CUDA Availability

```python
import torch

print('🔥 CUDA Status:')
print('='*70)

cuda_available = torch.cuda.is_available()
print(f'\nCUDA Available: {"✅ YES" if cuda_available else "❌ NO"}')

if cuda_available:
    print(f'\n📊 GPU Information:')
    print(f'   CUDA Version: {torch.version.cuda}')
    print(f'   Number of GPUs: {torch.cuda.device_count()}')
    print(f'   Current GPU: {torch.cuda.current_device()}')
    print(f'   GPU Name: {torch.cuda.get_device_name(0)}')
    
    props = torch.cuda.get_device_properties(0)
    print(f'\n💾 GPU Memory:')
    print(f'   Total: {props.total_memory / 1e9:.2f} GB')
    print(f'   Allocated: {torch.cuda.memory_allocated(0) / 1e9:.3f} GB')
    print(f'   Reserved: {torch.cuda.memory_reserved(0) / 1e9:.3f} GB')
    print(f'   Available: {(props.total_memory - torch.cuda.memory_allocated(0)) / 1e9:.2f} GB')
    
    print(f'\n⚡ GPU Specs:')
    print(f'   Compute Capability: {props.major}.{props.minor}')
    print(f'   Multi-processors: {props.multi_processor_count}')
else:
    print('\n⚠️  CUDA not available!')
    print('\n🔍 Possible reasons:')
    print('   1. PyTorch CPU-only version installed (torch.__version__ shows "+cpu")')
    print('   2. No NVIDIA GPU detected')
    print('   3. NVIDIA drivers not installed')
    print('\n💡 Solution:')
    print('   Reinstall PyTorch with CUDA:')
    print('   pip uninstall torch torchvision -y')
    print('   pip install torch torchvision --index-url https://download.pytorch.org/whl/cu121')

print('='*70)
```

## 🧪 Step 3: Test GPU Performance

```python
import torch
import time

print('🧪 GPU Performance Test:')
print('='*70)

if torch.cuda.is_available():
    size = 2000
    print(f'\n📐 Matrix multiplication test ({size}x{size})\n')
    
    # CPU Test
    print('⏱️  CPU Test...')
    cpu_a = torch.randn(size, size)
    cpu_b = torch.randn(size, size)
    
    cpu_start = time.time()
    cpu_c = torch.matmul(cpu_a, cpu_b)
    cpu_time = time.time() - cpu_start
    print(f'   CPU Time: {cpu_time:.4f}s')
    
    # GPU Test
    print('\n⏱️  GPU Test...')
    gpu_a = torch.randn(size, size).cuda()
    gpu_b = torch.randn(size, size).cuda()
    
    # Warmup
    _ = torch.matmul(gpu_a, gpu_b)
    torch.cuda.synchronize()
    
    gpu_start = time.time()
    gpu_c = torch.matmul(gpu_a, gpu_b)
    torch.cuda.synchronize()
    gpu_time = time.time() - gpu_start
    print(f'   GPU Time: {gpu_time:.4f}s')
    
    # Results
    speedup = cpu_time / gpu_time
    print(f'\n🚀 Results:')
    print(f'   GPU is {speedup:.2f}x faster than CPU')
    
    if speedup > 5:
        print('   ✅ GPU acceleration working perfectly!')
    elif speedup > 2:
        print('   ✅ GPU acceleration working (moderate speedup)')
    else:
        print('   ⚠️  GPU speedup lower than expected')
        print('   Check if GPU is being throttled or in power-saving mode')
    
    # Verify correctness using RELATIVE error
    abs_diff = (cpu_c - gpu_c.cpu()).abs().max().item()
    max_val = cpu_c.abs().max().item()
    rel_error = abs_diff / max_val if max_val > 0 else 0
    
    print(f'\n🔍 Accuracy Check:')
    print(f'   Max absolute difference: {abs_diff:.2e}')
    print(f'   Max value in result: {max_val:.2e}')
    print(f'   Relative error: {rel_error:.2e}')
    
    if rel_error < 1e-5:
        print('   ✅ Results match perfectly - GPU computing correctly!')
    elif rel_error < 1e-3:
        print('   ✅ Results match within expected precision - GPU working correctly!')
    else:
        print('   ⚠️  Large relative error - potential GPU issue')
else:
    print('⚠️  CUDA not available - skipping GPU test')
    print('   Install GPU-enabled PyTorch first')

print('='*70)
```

## 🧠 Step 4: Test Model Inference on GPU

```python
import torch
import torch.nn as nn
import time

print('🧠 Model Inference Test:')
print('='*70)

if torch.cuda.is_available():
    # Create a simple CNN model
    class SimpleModel(nn.Module):
        def __init__(self):
            super().__init__()
            self.conv1 = nn.Conv2d(3, 64, 3, padding=1)
            self.conv2 = nn.Conv2d(64, 128, 3, padding=1)
            self.conv3 = nn.Conv2d(128, 256, 3, padding=1)
            self.relu = nn.ReLU()
        
        def forward(self, x):
            x = self.relu(self.conv1(x))
            x = self.relu(self.conv2(x))
            x = self.relu(self.conv3(x))
            return x
    
    print('\n📦 Creating test model...')
    model = SimpleModel()
    print(f'   Parameters: {sum(p.numel() for p in model.parameters()):,}')
    
    # Test on CPU
    print('\n⏱️  CPU Inference:')
    model_cpu = model.cpu()
    input_cpu = torch.randn(1, 3, 512, 512)
    
    cpu_start = time.time()
    with torch.no_grad():
        output_cpu = model_cpu(input_cpu)
    cpu_time = time.time() - cpu_start
    print(f'   Time: {cpu_time:.4f}s')
    print(f'   Output shape: {output_cpu.shape}')
    
    # Test on GPU
    print('\n⏱️  GPU Inference:')
    model_gpu = model.cuda()
    input_gpu = torch.randn(1, 3, 512, 512).cuda()
    
    # Warmup
    with torch.no_grad():
        _ = model_gpu(input_gpu)
    torch.cuda.synchronize()
    
    gpu_start = time.time()
    with torch.no_grad():
        output_gpu = model_gpu(input_gpu)
    torch.cuda.synchronize()
    gpu_time = time.time() - gpu_start
    print(f'   Time: {gpu_time:.4f}s')
    print(f'   Output shape: {output_gpu.shape}')
    
    # Results
    speedup = cpu_time / gpu_time
    print(f'\n🚀 Model Inference Speedup: {speedup:.2f}x')
    
    print(f'\n💡 What this means for your models:')
    print(f'   CSRNet on GPU: ~{0.5 / speedup:.3f}s per image')
    print(f'   TMTB on GPU: ~{2.0 / speedup:.3f}s per image')
    
    if speedup > 3:
        print('\n✅ GPU acceleration is excellent!')
        print('   Your models will run much faster on GPU')
    else:
        print('\n⚠️  GPU speedup is lower than expected')
        print('   Models will still work, but performance may not be optimal')
else:
    print('⚠️  CUDA not available - skipping model test')

print('='*70)
```
